```python
import os
import random

# 1. 定義劃分比例 (加起來必須等於 1.0)
train_percent = 0.8
val_percent = 0.1
# 剩下的 0.1 自然就是 test

# 2. 定義路徑
txtfilepath = 'labels'
imgfilepath = './JPEGImages/'

# 3. 獲取所有標籤檔案
total_txt = os.listdir(txtfilepath)
num = len(total_txt)

# 4. 建立索引列表並隨機打亂 (Shuffling)
indices = list(range(num))
random.shuffle(indices)

# 5. 計算各個資料集的切點數量
train_count = int(num * train_percent)
val_count = int(num * val_percent)

# 6. 使用切片分配索引，保證彼此不重疊
train_indices = indices[:train_count]
val_indices = indices[train_count : train_count + val_count]
test_indices = indices[train_count + val_count :]

# 7. 只開啟 3 個檔案進行寫入
with open('train.txt', 'w') as ftrain, \
     open('val.txt', 'w') as fval, \
     open('test.txt', 'w') as ftest:

    # 效能優化：直接依據劃分好的索引寫入對應檔案
    for i in train_indices:
        img_name = imgfilepath + total_txt[i][:-4] + '.jpg\n'
        ftrain.write(img_name)

    for i in val_indices:
        img_name = imgfilepath + total_txt[i][:-4] + '.jpg\n'
        fval.write(img_name)

    for i in test_indices:
        img_name = imgfilepath + total_txt[i][:-4] + '.jpg\n'
        ftest.write(img_name)

print(f"資料劃分完成！總數: {num}")
print(f"訓練集 (train.txt) : {len(train_indices)} 筆")
print(f"驗證集 (val.txt)   : {len(val_indices)} 筆")
print(f"測試集 (test.txt)  : {len(test_indices)} 筆")
```
